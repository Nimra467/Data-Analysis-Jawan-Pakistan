GROUP BY --- HAVING

GROUP BY works after the WHERE clause filters the rows and before the HAVING clause filters the groups.

- Count Employees by JobTitle
- Average TotalPay by JobTitle
- Employees with the Highest TotalPay in Each Year
- Employees with TotalPay Greater Than 1,000,000 by JobTitle
- Employees with Average BasePay Greater Than 50,000 by JobTitle
- Employee Count in Each Year

```python
#Count Employees by JobTitle
q = '''select count(employeename) 'Employee Count', Jobtitle from 
    salaries group by jobtitle'''

sql_query(q)
```

```python
#Count Employees by JobTitle
q = '''select count(employeename) 'Employee Count', Jobtitle from 
    salaries 
    group by jobtitle
    where jobtitle != 'Not provided'
    '''

sql_query(q)
```

```python
#Count Employees by JobTitle
q = '''select count(employeename) 'Employee Count', Jobtitle from 
    salaries
    where jobtitle != 'Not provided'
    group by jobtitle
    order by 1
    '''

sql_query(q)
```

```python
#Count Employees by JobTitle
q = '''select count(employeename) 'Employee Count', Jobtitle from 
    salaries
    where jobtitle != 'Not provided'
    group by jobtitle
    order by 1 desc
    limit 10
    '''

sql_query(q)
```

```python
#Count Employees by JobTitle
q = '''select count(employeename) 'Employee Count', Jobtitle from 
    salaries
    where jobtitle != 'Not provided'
    group by jobtitle
    order by 1 desc
    limit 10
    '''

x = sql_query(q)
x.plot(kind='bar', x='JobTitle', y='Employee Count', color='red', figsize=(10,6))
```

```python
#Count Employees by JobTitle
q = '''select count(employeename) 'Employee Count', Jobtitle from 
    salaries
    where jobtitle != 'Not provided'
    group by jobtitle
    order by 1 desc
    limit 10
    '''

x = sql_query(q)
x.plot(kind='barh', x='JobTitle', y='Employee Count', color='green', figsize=(10,6))
```

### case

Categorizing Employees Based on overtimepay > 100000 'overtime worker' else 'regular worker'

```python
q = '''select employeename, overtimepay,
    case
    when overtimepay > 100000 then "overtime worker"
    else "regular worker"
    end as "work category"
    from salaries'''

sql_query(q)
    
    
```

Categorizing Job Titles Based on totalpay > 100000 'executive level', > 50k and < 99999 'management level' , else 'officer level'

```python
q = '''select jobtitle, totalpay,
    case
    when totalpay > 100000 then 'executive level'
    when totalpay between 50000 and 99999 then 'management level'
    else 'officer level'
    end as 'work category'
    from salaries
    '''

sql_query(q)
```

```python
q = '''select jobtitle, totalpay,
    case
    when totalpay > 100000 then 'executive level'
    when totalpay between 50000 and 99999 then 'management level'
    else 'officer level'
    end as 'work category'
    from salaries
    where jobtitle != 'Not provided'
    
    '''

sql_query(q)
```

```python
q = '''select jobtitle, totalpay,
    case
    when totalpay > 100000 then 'executive level'
    when totalpay between 50000 and 99999 then 'management level'
    else 'officer level'
    end as 'work category'
    from salaries
    where jobtitle != 'Not provided'
    limit 100
    '''

sql_query(q)
```

Find which year has missing data of Employees with 'Not Provided'/ 'Not provided'

```python
q='''SELECT DISTINCT Year
FROM salaries
WHERE JobTitle IN ('Not Provided', 'Not provided');
'''
sql_query(q)
```

Number of unique agencies

```python
q = '''SELECT COUNT(DISTINCT Agency) AS UniqueAgencies
FROM salaries;
'''
sql_query(q)
```

Show Average BasePay, OvertimePay, and TotalPay for year 2011

```python
q= '''SELECT 
    AVG(BasePay) AS AvgBasePay,
    AVG(OvertimePay) AS AvgOvertimePay,
    AVG(TotalPay) AS AvgTotalPay
FROM salaries
WHERE Year = 2011;
'''
sql_query(q)
```

Plot the top 10 job titles with the highest TotalPay in 2014.


```python
q = '''SELECT JobTitle, SUM(TotalPay) AS TotalPaySum
FROM salaries
WHERE Year = 2014
GROUP BY JobTitle
ORDER BY TotalPaySum DESC
LIMIT 10;
'''
sql_query(q)
```

Which job titles have overtime pay exceeding $150,000 exluding 'Not provided'

```python
q = '''SELECT DISTINCT JobTitle
FROM salaries
WHERE OvertimePay > 150000
AND JobTitle NOT IN ('Not Provided', 'Not provided');
'''
sql_query(q)
```

Find all employees whose job titles contain the word “CHIEF”

```python
q = '''SELECT EmployeeName, JobTitle
FROM salaries
WHERE JobTitle LIKE '%CHIEF%';
'''
sql_query(q)
```

Count how many employees have job titles containing “ENGINEER” and “Fire” in year 2012 and 2013

```python
q = '''SELECT Year, COUNT(*) AS EmployeeCount
FROM salaries
WHERE Year IN (2012, 2013)
AND (JobTitle LIKE '%ENGINEER%' OR JobTitle LIKE '%Fire%')
GROUP BY Year;
'''
sql_query(q)
```

Employees having basepay > 50k and totalpay < 150000 exlucding 'Not Provided'
